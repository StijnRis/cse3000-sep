```python
import pandas as pd
import json
from graphs.all_over_time import *

with open('categories.json') as f:
    categories_data = json.load(f)
category_order = []
for mapping in categories_data:
    category_order.extend(mapping.get('categories', []))
```

```python
DATASOURCES = ['zeroshot', 'llm']
CONFIDENCE_THRESHOLD = 0.80

for DATASOURCE in DATASOURCES:
    print(f"Processing data for DATASOURCE: {DATASOURCE}")

    df_domain_total = pd.read_csv(f"results/{DATASOURCE}_{CONFIDENCE_THRESHOLD}_domain_total.csv")
    df_category_total = pd.read_csv(f"results/{DATASOURCE}_{CONFIDENCE_THRESHOLD}_category_total.csv")

    df_year_summed_average_per_year = pd.read_csv(f"results/{DATASOURCE}_{CONFIDENCE_THRESHOLD}_summed_results_per_year.csv")
    df_year_final_summed = pd.read_csv(f"results/{DATASOURCE}_{CONFIDENCE_THRESHOLD}_year_final_summed.csv")

    create_domain_comparison(df_domain_total, f"{DATASOURCE}_{CONFIDENCE_THRESHOLD}")
    print("Domain comparison chart saved.")

    create_category_comparison(df_category_total, category_order, f"{DATASOURCE}_{CONFIDENCE_THRESHOLD}")
    print("Category comparison chart saved.")

    create_category_trends_over_time(df_year_summed_average_per_year, category_order, f"{DATASOURCE}_{CONFIDENCE_THRESHOLD}")
    print("Category Trends Over Time chart saved.")

    # Create category distribution graph per project
    for project in df_year_final_summed['project_name'].unique():
        df_project = df_year_final_summed[df_year_final_summed['project_name'] == project]
        create_category_distribution_graph(df_project, category_order, project, path=f'{DATASOURCE}_{CONFIDENCE_THRESHOLD}')
        print(f"Category distribution graph for {project} saved.")

df_year_summed_average_zeroshot = pd.read_csv(f"results/zeroshot_{CONFIDENCE_THRESHOLD}_summed_results.csv")
df_year_summed_average_llm = pd.read_csv(f"results/llm_{CONFIDENCE_THRESHOLD}_summed_results.csv")

create_grouped_bar_chart(df_year_summed_average_zeroshot, df_year_summed_average_llm, category_order, f"{CONFIDENCE_THRESHOLD}")
```
